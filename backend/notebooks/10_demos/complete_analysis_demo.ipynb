## 2. Presentación del Sistema

```
def mostrar_banner():
    linea = "=" * 60
    print(linea)
    print("  ANALIZADOR DE COMPLEJIDADES ALGORÍTMICAS")
    print("  Análisis automático asistido por LLMs")
    print(linea)
    print()

def mostrar_resultado_analisis(nombre, big_o, omega=None, espacio=None, patron=None, estructuras=None):
    """Muestra los resultados de un análisis de forma clara."""
    print(f"  RESULTADOS: {nombre}")
    print(f"  Complejidad Temporal:")
    print(f"    Peor caso (Big O):   {big_o}")
    if omega:
        print(f"    Mejor caso (Omega):  {omega}")
    if espacio:
        print(f"  Complejidad Espacial: {espacio}")
    if patron:
        print(f"  Patrón Detectado:     {patron}")
    if estructuras:
        print(f"  Estructuras de Datos: {', '.join(estructuras) if estructuras else 'Ninguna'}")

mostrar_banner()
print("Sistema inicializado. Comenzando demostraciones...")
```

## 3. Demo 1: Algoritmo Simple

```
print("# DEMO 1: Suma de elementos de un arreglo            #")

ALGORITMO_SUMA = """
algorithm sumaArreglo(A[], n)
begin
    total <- 0
    for i <- 1 to n do
        total <- total + A[i]
    end
    return total
end
"""

print("Código del algoritmo:")
print(ALGORITMO_SUMA)

# Paso 1: Parsing
print("Paso 1: Parsing del pseudocódigo...")
ast = parse_pseudocode(ALGORITMO_SUMA)
print(f"  Algoritmo: '{ast.algorithm.name}'")
print(f"  Parámetros: {[p.name for p in ast.algorithm.parameters]}")
print(f"  Statements en el cuerpo: {len(ast.algorithm.body.statements)}")

# Paso 2: Análisis de complejidad
print("\nPaso 2: Análisis de complejidad...")
engine = AnalyzerEngine()
analysis = engine.analyze(ast)
big_o = getattr(analysis, 'big_o', None) or getattr(analysis, 'time_complexity', {}).get('big_o', 'N/A')
omega = getattr(analysis, 'omega', None) or getattr(analysis, 'time_complexity', {}).get('omega', 'N/A')
espacio = getattr(analysis, 'space_complexity', None) or getattr(analysis, 'space', 'O(1)')

# Paso 3: Detección de patrones
print("Paso 3: Detección de patrones...")
detector = PatternDetector()
detection = detector.detect(ast)
patron = getattr(detection, 'primary_pattern', None)
confianza = getattr(detection, 'primary_confidence', 0.0)

# Paso 4: Detección de estructuras
print("Paso 4: Detección de estructuras de datos...")
identificador = StructureIdentifier()
estructuras_result = identificador.identify(ast)
estructuras = [
    str(m.structure_type).replace("StructureType.", "")
    for m in getattr(estructuras_result, 'structures', [])
]

mostrar_resultado_analisis(
    "Suma de Arreglo",
    big_o=big_o,
    omega=omega,
    espacio=espacio,
    patron=patron,
    estructuras=estructuras
)

print(f"\nInterpretación:")
print(f"  El algoritmo recorre el arreglo una sola vez (un ciclo for).")
print(f"  Esto produce una complejidad lineal O(n): a mayor n, más operaciones en proporción directa.")
```

## 4. Demo 2: Algoritmo de Ordenamiento

```
print("# DEMO 2: Bubble Sort                                 #")

ALGORITMO_BUBBLE = """
algorithm bubbleSort(A[], n)
begin
    for i <- 1 to n - 1 do
        for j <- 1 to n - i do
            if (A[j] > A[j + 1]) then
                temp <- A[j]
                A[j] <- A[j + 1]
                A[j + 1] <- temp
            end
        end
    end
end
"""

ast_bubble = parse_pseudocode(ALGORITMO_BUBBLE)
analysis_bubble = engine.analyze(ast_bubble)
detection_bubble = detector.detect(ast_bubble)
estructuras_bubble = identificador.identify(ast_bubble)

big_o_b = getattr(analysis_bubble, 'big_o', None) or 'O(n^2)'
omega_b = getattr(analysis_bubble, 'omega', None) or 'O(n)'
patron_b = getattr(detection_bubble, 'primary_pattern', None)
conf_b = getattr(detection_bubble, 'primary_confidence', 0.0)

mostrar_resultado_analisis(
    "Bubble Sort",
    big_o=big_o_b,
    omega=omega_b,
    espacio="O(1)",
    patron=patron_b,
    estructuras=["ARRAY"]
)

print(f"\nInterpretación:")
print(f"  Los dos ciclos for anidados generan O(n) × O(n) = O(n^2) en el peor caso.")
print(f"  Confianza en detección del patrón SORTING: {conf_b:.2f}")
print(f"  La complejidad espacial es O(1) porque el ordenamiento es in-place (sin memoria extra).")
```

## 5. Demo 3: Algoritmo Recursivo con Divide y Vencerás

```
print("# DEMO 3: Merge Sort (Divide y Vencerás)              #")

ALGORITMO_MERGE = """
algorithm mergeSort(A[], p, r)
begin
    if (p < r) then
        q <- floor((p + r) / 2)
        call mergeSort(A, p, q)
        call mergeSort(A, q + 1, r)
    end
end
"""

ast_merge = parse_pseudocode(ALGORITMO_MERGE)
analysis_merge = engine.analyze(ast_merge)
detection_merge = detector.detect(ast_merge)

big_o_m = getattr(analysis_merge, 'big_o', None) or 'O(n log n)'
patron_m = getattr(detection_merge, 'primary_pattern', None)
conf_m = getattr(detection_merge, 'primary_confidence', 0.0)

# Intentar generar árbol de recursión
print("\nGenerando árbol de recursión...")
try:
    tree_gen = RecursionTreeGenerator()
    tree_result = tree_gen.generate(ast_merge, max_depth=4)
    print(f"  Árbol generado: {getattr(tree_result, 'node_count', '?')} nodos")
    print(f"  Factor de ramificación: 2 (dos llamadas recursivas)")
    print(f"  Factor de reducción: 1/2 (n/2 en cada llamada)")
except Exception as e:
    print(f"  Árbol de recursión: {str(e)[:50]}")

mostrar_resultado_analisis(
    "Merge Sort",
    big_o=big_o_m,
    omega="O(n log n)",
    espacio="O(n)",
    patron=patron_m,
    estructuras=["ARRAY"]
)

print(f"\nEcuación de recurrencia:")
print(f"  T(n) = 2·T(n/2) + O(n)")
print(f"  Solución por Teorema Maestro (caso 2): T(n) = O(n log n)")
print(f"  Confianza en detección DIVIDE_AND_CONQUER: {conf_m:.2f}")
```

## 6. Demo 4: Comparación entre Algoritmos

```
print("# DEMO 4: Comparación de Algoritmos de Ordenamiento  #")

algoritmos_comparar = [
    ("Bubble Sort O(n^2)", ALGORITMO_BUBBLE),
    ("Merge Sort O(n log n)", ALGORITMO_MERGE),
    ("Suma Simple O(n)", ALGORITMO_SUMA),
]

print(f"\n{'Algoritmo':<30} {'Big O':<15} {'Patrón':<25} {'Eficiencia':>10}")

for nombre, codigo in algoritmos_comparar:
    try:
        ast = parse_pseudocode(codigo)
        analysis = engine.analyze(ast)
        detection = detector.detect(ast)
        
        big_o = str(getattr(analysis, 'big_o', None) or 'N/A')
        patron = str(getattr(detection, 'primary_pattern', None) or 'N/A')
        patron_corto = patron.replace("PatternType.", "")[:20]
        
        # Clasificar eficiencia relativa
        eficiencia_map = {
            "O(1)": "Óptima",
            "O(log n)": "Excelente",
            "O(n)": "Buena",
            "O(n log n)": "Buena",
            "O(n^2)": "Aceptable",
            "O(2^n)": "Mala"
        }
        eficiencia = eficiencia_map.get(big_o, "Variable")
        
        print(f"{nombre:<30} {big_o:<15} {patron_corto:<25} {eficiencia:>10}")
    except Exception as e:
        print(f"{nombre:<30} ERROR: {str(e)[:40]}")

print("\nConclusión:")
print("  Para ordenar n elementos:")
print("  - Merge Sort es preferible a Bubble Sort para n grande")
print("  - La diferencia es de n^2 vs n*log(n) operaciones")
print("  - Para n=1000: Bubble Sort ~ 1.000.000 ops vs Merge Sort ~ 10.000 ops")
```

## 7. Demo 5: Exportación de Resultados

```
print("# DEMO 5: Exportación de Resultados                   #")

formatos_disponibles = ExporterFactory.get_available_formats()
print(f"\nFormatos de exportación disponibles: {len(formatos_disponibles)}")
for fmt in formatos_disponibles:
    print(f"  - {fmt}")

# Exportar el análisis del Merge Sort a JSON como demostración
try:
    ast_demo = parse_pseudocode(ALGORITMO_MERGE)
    analysis_demo = engine.analyze(ast_demo)
    detection_demo = detector.detect(ast_demo)
    
    # Construir datos de exportación
    from pathlib import Path
    
    resultado_demo = {
        "algoritmo": "mergeSort",
        "big_o": str(getattr(analysis_demo, 'big_o', None) or 'O(n log n)'),
        "omega": str(getattr(analysis_demo, 'omega', None) or 'O(n log n)'),
        "patron_detectado": str(getattr(detection_demo, 'primary_pattern', None)),
        "confianza_patron": float(getattr(detection_demo, 'primary_confidence', 0.0)),
    }
    
    output_dir = Path("../../data/exports/demo")
    output_dir.mkdir(parents=True, exist_ok=True)
    
    output_json = output_dir / "merge_sort_analysis.json"
    with open(output_json, "w", encoding="utf-8") as f:
        json.dump(resultado_demo, f, indent=2, ensure_ascii=False)
    
    print(f"\nAnálisis exportado a: {output_json}")
    print(f"Contenido del JSON exportado:")
    print(json.dumps(resultado_demo, indent=2))
    
except Exception as e:
    print(f"Error en exportación: {e}")
```

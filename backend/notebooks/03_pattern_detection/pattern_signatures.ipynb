## 3. Indicadores por Patron

### Indicadores de Divide y Venceras

| Indicador | Peso | Descripcion |
|-----------|------|-------------|
| calculo_medio | 1.5 | mid = (low+high)/2 |
| multiples_llamadas | 2.0 | 2+ llamadas recursivas |
| fase_combinacion | 1.0 | Llamada a merge |

### Indicadores de Programacion Dinamica

| Indicador | Peso | Descripcion |
|-----------|------|-------------|
| tabla_memoizacion | 2.0 | dp[], memo[] |
| acceso_anterior | 1.5 | dp[i-1] |

### Ejemplo: Detectar Divide y Venceras

```python
# Detectar Divide y Venceras: Merge Sort
codigo_merge = """
algorithm mergeSort(A[], p, r)
begin
    if (p < r) then
        q <- floor((p + r) / 2)
        call mergeSort(A, p, q)
        call mergeSort(A, q + 1, r)
        call merge(A, p, q, r)
    end
end
"""

ast = parse_pseudocode(codigo_merge)
resultado = detector.detect(ast)

print("=== MERGE SORT ===")
print(f"Patron primario: {resultado.primary_pattern_name}")
print(f"Confianza: {resultado.primary_confidence:.2%}")

# Mostrar indicadores detectados
if resultado.all_patterns:
    primary = resultado.all_patterns[0]
    print(f"\nIndicadores encontrados:")
    for indicator in primary.pattern.indicators:
        status = "ENCONTRADO" if indicator.found else "no encontrado"
        print(f"  - {indicator.name}: {status}")
```

### Ejemplo: Detectar Programacion Dinamica

```python
# Detectar Programacion Dinamica: Fibonacci
codigo_dp = """
algorithm fibonacciDP(n)
begin
    dp[0] <- 0
    dp[1] <- 1
    for i <- 2 to n do
        dp[i] <- dp[i - 1] + dp[i - 2]
    end
    return dp[n]
end
"""

ast = parse_pseudocode(codigo_dp)
resultado = detector.detect(ast)

print("=== FIBONACCI DP ===")
print(f"Patron primario: {resultado.primary_pattern_name}")
print(f"Confianza: {resultado.primary_confidence:.2%}")
```

## 4. Deteccion de Multiples Patrones

Un algoritmo puede exhibir caracteristicas de multiples patrones.
El sistema detecta todos y asigna un patron primario.

### Ejemplo: Comparar Varios Algoritmos

```python
# Comparar deteccion en varios algoritmos
algoritmos = {
    "Bubble Sort": """
algorithm bubbleSort(A[], n)
begin
    for i <- 1 to n - 1 do
        for j <- 1 to n - i do
            if (A[j] > A[j + 1]) then
                temp <- A[j]
                A[j] <- A[j + 1]
                A[j + 1] <- temp
            end
        end
    end
end
""",
    "Binary Search": """
algorithm binarySearch(A[], low, high, key)
begin
    if (low <= high) then
        mid <- floor((low + high) / 2)
        if (A[mid] = key) then
            return mid
        end
        if (A[mid] > key) then
            return call binarySearch(A, low, mid - 1, key)
        else
            return call binarySearch(A, mid + 1, high, key)
        end
    end
    return -1
end
""",
    "Factorial": """
algorithm factorial(n)
begin
    if (n <= 1) then
        return 1
    end
    return n * call factorial(n - 1)
end
"""
}

resultados = []
for nombre, codigo in algoritmos.items():
    ast = parse_pseudocode(codigo)
    resultado = detector.detect(ast)
    resultados.append({
        "Algoritmo": nombre,
        "Patron Primario": resultado.primary_pattern_name,
        "Confianza": f"{resultado.primary_confidence:.0%}",
        "Patrones Detectados": resultado.pattern_count
    })

df = pd.DataFrame(resultados)
print("=== COMPARACION DE PATRONES ===")
print(df.to_string(index=False))
```

## 5. Ejercicios

### Ejercicio 1: Identificar el Patron

```python
# Ejercicio 1: Que patron tiene este algoritmo?
# Predice antes de ejecutar!

codigo_ejercicio = """
algorithm sumArray(A[], n)
begin
    sum <- 0
    for i <- 1 to n do
        sum <- sum + A[i]
    end
    return sum
end
"""

# Tu prediccion: ________________

ast = parse_pseudocode(codigo_ejercicio)
resultado = detector.detect(ast)
print(f"Patron detectado: {resultado.primary_pattern_name}")
print(f"Confianza: {resultado.primary_confidence:.0%}")
```

### Ejercicio 2: Escribir un Patron Especifico

```python
# Ejercicio 2: Escribe un algoritmo que sea detectado como RECURSIVO
# Pista: debe tener una llamada a si mismo y un caso base

tu_codigo = """
algorithm tuAlgoritmo(n)
begin
    if (n <= 0) then
        return 0
    end
    return 1 + call tuAlgoritmo(n - 1)
end
"""

ast = parse_pseudocode(tu_codigo)
resultado = detector.detect(ast)

print(f"Patron detectado: {resultado.primary_pattern_name}")
es_recursivo = any(p.pattern.pattern_type == PatternType.RECURSIVE 
                   for p in resultado.all_patterns)
print(f"Es recursivo: {'Si!' if es_recursivo else 'No'}")
```

## Introduccion a los Patrones

Los patrones algoritmicos son estrategias de diseno que resuelven problemas
computacionales de manera caracteristica. El sistema detecta estos patrones
analizando la estructura del AST.

### Patrones Soportados

| Patron | Descripcion | Complejidad Tipica |
|--------|-------------|--------------------|
| Fuerza Bruta | Prueba todas las soluciones | O(n^k), O(2^n) |
| Recursivo | Solucion que se llama a si misma | Variable |
| Divide y Venceras | Divide, resuelve, combina | O(n log n) |
| Programacion Dinamica | Subproblemas con memoizacion | O(n^2), O(nW) |
| Voraz (Greedy) | Eleccion localmente optima | O(n log n) |
| Backtracking | Exploracion con retroceso | O(k^n), O(n!) |
| Branch and Bound | Backtracking con poda | Variable |
| Ordenamiento | Algoritmos de ordenar | O(n^2), O(n log n) |
| Busqueda | Algoritmos de buscar | O(n), O(log n) |

## Patrones Implementados

### 1. Fuerza Bruta (Brute Force)

**Archivo**: `detectors/brute_force.py`

**Caracteristicas**:
- Multiples ciclos anidados
- Sin optimizacion de espacio de busqueda
- Examina todas las posibilidades

**Ejemplo tipico**:

```
algorithm bruteForceSubset(A[], n, target)
begin
    for i <- 0 to 2^n - 1 do
        sum <- 0
        for j <- 0 to n - 1 do
            if (bit(i, j) = 1) then
                sum <- sum + A[j]
            end
        end
        if (sum = target) then
            return true
        end
    end
    return false
end
```

### 2. Recursivo

**Archivo**: `detectors/recursive_detector.py`

**Caracteristicas**:
- Llamadas a si mismo (call nombreAlgoritmo)
- Caso base con condicion de parada
- Reduccion del problema

**Ejemplo tipico**:

```
algorithm factorial(n)
begin
    if (n <= 1) then
        return 1
    else
        return n * call factorial(n - 1)
    end
end
```

### 3. Divide y Venceras

**Archivo**: `detectors/divide_conquer_detector.py`

**Caracteristicas**:
- Division del problema en subproblemas
- Llamadas recursivas con parametros reducidos (n/2)
- Fase de combinacion (merge)

**Indicadores clave**:
- Calculo de punto medio: `mid <- (low + high) / 2`
- Dos o mas llamadas recursivas
- Paso de combinacion de resultados

**Ejemplo tipico**:

```
algorithm mergeSort(A[], p, r)
begin
    if (p < r) then
        q <- floor((p + r) / 2)
        call mergeSort(A, p, q)
        call mergeSort(A, q + 1, r)
        call merge(A, p, q, r)
    end
end
```

### 4. Programacion Dinamica

**Archivo**: `detectors/dynamic_programming_detector.py`

**Caracteristicas**:
- Tabla de memoizacion (array 1D o 2D)
- Construccion bottom-up o top-down
- Acceso a resultados previos: `dp[i-1]`, `memo[n]`

**Indicadores clave**:
- Declaracion de tabla: `dp[n]`, `memo[n][m]`
- Referencias a posiciones anteriores
- Patron de llenado de tabla

**Ejemplo tipico**:

```
algorithm fibonacciDP(n)
begin
    dp[0] <- 0
    dp[1] <- 1
    for i <- 2 to n do
        dp[i] <- dp[i-1] + dp[i-2]
    end
    return dp[n]
end
```

### 5. Voraz (Greedy)

**Archivo**: `detectors/greedy_detector.py`

**Caracteristicas**:
- Eleccion localmente optima en cada paso
- Sin reconsideracion de decisiones
- Construccion incremental de solucion

**Indicadores clave**:
- Ordenamiento inicial
- Seleccion del mejor candidato
- Adicion a solucion parcial

**Ejemplo tipico**:

```
algorithm activitySelection(s[], f[], n)
begin
    // Ordenar por tiempo de finalizacion
    call sort(activities, by: finish_time)
    
    selected <- 1
    last <- 1
    
    for i <- 2 to n do
        if (s[i] >= f[last]) then
            selected <- selected + 1
            last <- i
        end
    end
    
    return selected
end
```

### 6. Backtracking

**Archivo**: `detectors/backtracking_detector.py`

**Caracteristicas**:
- Exploracion sistematica
- Retroceso cuando no hay solucion
- Poda del espacio de busqueda

**Indicadores clave**:
- Verificacion de validez antes de continuar
- Deshacer cambios (undo)
- Recursion con estado mutable

**Ejemplo tipico**:

```
algorithm nQueens(board[][], row, n)
begin
    if (row = n) then
        return true
    end
    
    for col <- 0 to n - 1 do
        if (isSafe(board, row, col, n)) then
            board[row][col] <- 1
            if (call nQueens(board, row + 1, n)) then
                return true
            end
            board[row][col] <- 0  // Backtrack
        end
    end
    
    return false
end
```

### 7. Branch and Bound

**Archivo**: `detectors/branch_bound_detector.py`

**Caracteristicas**:
- Backtracking con evaluacion de limites
- Poda basada en cota inferior/superior
- Mantiene mejor solucion encontrada

**Indicadores clave**:
- Variable `best` o `optimo`
- Comparacion con cota antes de explorar
- Funcion de bound/cota

### 8. Ordenamiento

**Archivo**: `detectors/sorting_and_searching.py`

**Subtipos detectados**:
- Bubble Sort: intercambios adyacentes
- Selection Sort: busqueda de minimo
- Insertion Sort: insercion en posicion
- Quick Sort: particionado
- Merge Sort: division y fusion

### 9. Busqueda

**Archivo**: `detectors/sorting_and_searching.py`

**Subtipos detectados**:
- Busqueda Lineal: recorrido secuencial
- Busqueda Binaria: division por mitad
- Busqueda en Arbol: navegacion jerarquica

## Indicadores por Patron

Cada patron se detecta mediante un conjunto de indicadores:

### Estructura de un Indicador

```python
@dataclass
class PatternIndicator:
    name: str              # "llamada_recursiva"
    description: str       # "Algoritmo se llama a si mismo"
    found: bool            # True si se encontro
    weight: float = 1.0    # Peso en el calculo de confianza
    evidence: str = None   # "Linea 5: call fibonacci(n-1)"
    location: str = None   # "ForLoopNode en linea 10"
```

### Indicadores de Divide y Venceras

| Indicador | Peso | Descripcion |
|-----------|------|-------------|
| calculo_medio | 1.5 | Calculo de punto medio |
| multiples_llamadas | 2.0 | Dos o mas llamadas recursivas |
| parametros_reducidos | 1.5 | Llamadas con n/2, p-q |
| fase_combinacion | 1.0 | Llamada a merge/combine |
| caso_base | 1.0 | Condicion de terminacion |

### Indicadores de Programacion Dinamica

| Indicador | Peso | Descripcion |
|-----------|------|-------------|
| tabla_memoizacion | 2.0 | Declaracion de dp/memo |
| acceso_anterior | 1.5 | Referencias a dp[i-1] |
| llenado_iterativo | 1.0 | Ciclo de construccion |
| inicializacion_base | 1.0 | dp[0] = valor |

### Indicadores de Backtracking

| Indicador | Peso | Descripcion |
|-----------|------|-------------|
| verificacion_validez | 1.5 | isSafe, isValid |
| deshacer_cambio | 2.0 | Asignacion despues de recursion |
| exploracion_opciones | 1.0 | Ciclo sobre candidatos |
| retorno_booleano | 1.0 | return true/false |

## Sistema de Deteccion

### Flujo de Deteccion

```
1. PatternDetector recibe AST
   |
2. Ejecuta todos los detectores
   |-- BruteForceDetector.detect(ast)
   |-- RecursiveDetector.detect(ast)
   |-- DivideConquerDetector.detect(ast)
   |-- ...
   |
3. Cada detector retorna PatternMatch
   |
4. PatternScorer rankea resultados
   |-- Ajusta scores
   |-- Detecta conflictos
   |-- Identifica patron primario
   |
5. Retorna PatternDetectionResult
```

### Resultado de Deteccion

```python
@dataclass
class PatternDetectionResult:
    primary_pattern: ScoredPattern         # Patron principal
    all_patterns: List[ScoredPattern]      # Todos los detectados
    confident_patterns: List[ScoredPattern] # Confianza >= umbral
    summary: str                           # Resumen textual
    metadata: Dict[str, Any]               # Info adicional
```

### Metodos Utiles

| Metodo | Descripcion |
|--------|-------------|
| `detect(ast)` | Detecta todos los patrones |
| `detect_specific(ast, type)` | Detecta un patron especifico |
| `has_patterns` | True si se detecto algo |
| `primary_pattern_name` | Nombre del patron principal |
| `primary_confidence` | Confianza del patron principal |

## Demostracion: Deteccion de Patrones

````python
import sys
sys.path.insert(0, '../..')

from app.core.parser.pseudocode_parser import PseudocodeParser
from app.core.patterns.pattern_detector import PatternDetector

parser = PseudocodeParser()
detector = PatternDetector()

# Ejemplo: Merge Sort (Divide y Venceras)
codigo = '''
algorithm mergeSort(A[], p, r)
begin
    if (p < r) then
        q <- floor((p + r) / 2)
        call mergeSort(A, p, q)
        call mergeSort(A, q + 1, r)
        call merge(A, p, q, r)
    end
end
'''

ast = parser.parse(codigo)
resultado = detector.detect(ast)

print(f"Algoritmo: {ast.algorithm.name}")
print(f"Patron primario: {resultado.primary_pattern_name}")
print(f"Confianza: {resultado.primary_confidence:.2%}")
print(f"Numero de patrones detectados: {resultado.pattern_count}")
print(f"\nResumen: {resultado.summary}")
````

## Conclusiones

El sistema de firmas de patrones:

1. **Detecta multiples patrones**: Un algoritmo puede exhibir varios patrones
2. **Utiliza indicadores ponderados**: Cada indicador tiene un peso
3. **Proporciona confianza**: Nivel de certeza en la deteccion
4. **Identifica patron primario**: El patron mas relevante

Los patrones detectados informan:
- Estrategia de resolucion utilizada
- Complejidad esperada
- Tecnicas de optimizacion posibles

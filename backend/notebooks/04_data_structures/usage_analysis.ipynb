## 3. Frecuencia de Operaciones

### 3.1 Ejemplo Básico

````
code_example = """
algorithm processArray(A[n])
begin
    sum ← 0
    
    for i ← 1 to n do
        sum ← sum + A[i]
    
    for i ← 1 to n do
        A[i] ← A[i] * 2
    
    for i ← 1 to n do
        print(A[i])
    
    return sum
end
"""

# Identificar y analizar
ast = parse_pseudocode(code_example)
result = identify_structures(ast)

if result.primary_structure:
    # Analizar uso
    usage = analyze_structure_usage(ast, result.primary_structure)
    
    print("ANÁLISIS DE USO")
    print(f"Estructura: {result.primary_structure.structure_name}")
    print(f"\nTotal operaciones: {usage.total_operations}")
    print(f"Operaciones únicas: {usage.unique_operations}")
    print(f"Operación dominante: {usage.most_frequent_operation}")
````

### 3.2 Tabla de Frecuencias

````
if usage.operation_frequencies:
    print("\nFRECUENCIA DE OPERACIONES:")
    
    freq_data = []
    for op_freq in usage.operation_frequencies:
        freq_data.append({
            "Operación": op_freq.operation,
            "Frecuencia": op_freq.count,
            "Complejidad": op_freq.complexity
        })
    
    df_freq = pd.DataFrame(freq_data)
    print(df_freq.to_string(index=False))
````

### 3.3 Gráfico de Frecuencias

````
if usage.operation_frequencies:
    # Preparar datos
    operations = [op.operation for op in usage.operation_frequencies]
    counts = [op.count for op in usage.operation_frequencies]
    
    # Crear gráfico
    plt.figure(figsize=(10, 6))
    plt.bar(operations, counts, color='steelblue', alpha=0.7)
    plt.xlabel('Operación')
    plt.ylabel('Frecuencia')
    plt.title('Frecuencia de Operaciones sobre la Estructura')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
````

## 4. Patrones de Acceso

### 4.1 Tipos de Patrones

| Patrón | Descripción | Típico en |
|--------|-------------|-----------|
| **sequential** | Acceso secuencial | Arrays con for loops |
| **random_access** | Acceso aleatorio | Arrays con índices variables |
| **sequential_lifo** | LIFO secuencial | Stacks |
| **sequential_fifo** | FIFO secuencial | Queues |
| **tree_traversal** | Recorrido de árbol | Trees |
| **graph_traversal** | Recorrido de grafo | Graphs |

### 4.2 Ejemplo: Patrón Sequential

````
code_sequential = """
algorithm sumArray(A[n])
begin
    sum ← 0
    for i ← 1 to n do
        sum ← sum + A[i]
    return sum
end
"""

ast = parse_pseudocode(code_sequential)
result = identify_structures(ast)
usage = analyze_structure_usage(ast, result.primary_structure)

print("PATRÓN DE ACCESO")
print(f"Patrón detectado: {usage.access_pattern}")
print(f"\nDescripción:")

pattern_descriptions = {
    "sequential": "Acceso secuencial óptimo para arrays. Buena localidad de caché.",
    "random_access": "Acceso aleatorio. Arrays son ideales para esto (O(1)).",
    "sequential_lifo": "Patrón LIFO. Stack apropiado para este caso.",
    "sequential_fifo": "Patrón FIFO. Queue apropiado para este caso."
}

print(f"  {pattern_descriptions.get(usage.access_pattern, 'Patrón mixto')}")
````

### 4.3 Ejemplo: Patrón Random Access

````
code_random = """
algorithm binarySearch(A[n], x)
begin
    left ← 1
    right ← n
    
    while left <= right do
    begin
        mid ← (left + right) / 2
        
        if A[mid] = x then
            return mid
        else if A[mid] < x then
            left ← mid + 1
        else
            right ← mid - 1
    end
    
    return -1
end
"""

ast = parse_pseudocode(code_random)
result = identify_structures(ast)
usage = analyze_structure_usage(ast, result.primary_structure)

print("\nPATRÓN: BINARY SEARCH")
print(f"Patrón: {usage.access_pattern}")
print(f"Operación dominante: {usage.most_frequent_operation}")
````

### 5.2 Comparación de Estructuras

Analiza el mismo algoritmo con diferentes estructuras:

````
# Implementación con array
code_with_array = """
algorithm findElement(data[n], target)
begin
    for i ← 1 to n do
        if data[i] = target then
            return true
    return false
end
"""

# Implementación con set (hipotético)
code_with_set = """
algorithm findElement(dataSet, target)
begin
    if target in dataSet then
        return true
    return false
end
"""

print("COMPARACIÓN DE IMPLEMENTACIONES")

# Array
ast_array = parse_pseudocode(code_with_array)
result_array = identify_structures(ast_array)
if result_array.primary_structure:
    usage_array = analyze_structure_usage(ast_array, result_array.primary_structure)
    print(f"\n Con Array:")
    print(f"   Operación dominante: {usage_array.most_frequent_operation}")
    if 'dominant_complexity' in usage_array.complexity_impact:
        print(f"   Complejidad: {usage_array.complexity_impact['dominant_complexity']}")

# Set (conceptual)
print(f"\nCon Set (conceptual):")
print(f"   Operación dominante: contains")
print(f"   Complejidad: O(1) promedio")
print(f"\nSet sería más eficiente para búsquedas frecuentes")
````

## 6. Recomendaciones

### 6.1 Ver Recomendaciones

````
code_with_searches = """
algorithm processData(list, queries[m])
begin
    results ← createArray(m)
    
    for i ← 1 to m do
    begin
        found ← false
        for item in list do
        begin
            if item = queries[i] then
            begin
                found ← true
                break
            end
        end
        results[i] ← found
    end
    
    return results
end
"""

ast = parse_pseudocode(code_with_searches)
result = identify_structures(ast)
usage = analyze_structure_usage(ast, result.primary_structure)

print("RECOMENDACIONES")

if usage.recommendations:
    for i, rec in enumerate(usage.recommendations, 1):
        print(f"\n{i}. {rec}")
else:
    print("Sin recomendaciones específicas")
````

### 6.2 Tipos de Recomendaciones

El analyzer puede generar:

1. **Optimización de Estructura**
   - "Considerar usar hash table para búsquedas frecuentes"
   - "Array es óptimo para acceso secuencial"

2. **Complejidad**
   - "Operación más costosa: O(n²). Considerar optimización"
   - "Estructura óptima para operaciones O(1)"

3. **Uso**
   - "Acceso secuencial óptimo para arrays"
   - "Stack apropiado para operaciones LIFO"

## 7. Caso de Estudio: Optimización

### 7.1 Problema Original

````
# Implementación no óptima: búsquedas repetidas en array
code_original = """
algorithm countOccurrences(data[n], queries[m])
begin
    results ← createArray(m)
    
    for i ← 1 to m do
    begin
        count ← 0
        for j ← 1 to n do
        begin
            if data[j] = queries[i] then
                count ← count + 1
        end
        results[i] ← count
    end
    
    return results
end
"""

print("ANÁLISIS: IMPLEMENTACIÓN ORIGINAL")

ast_orig = parse_pseudocode(code_original)
result_orig = identify_structures(ast_orig, min_confidence=0.25)

# Analizar todas las estructuras
for struct in result_orig.structures_found:
    usage = analyze_structure_usage(ast_orig, struct)
    
    print(f"\n{struct.structure_name}:")
    print(f"   Total operaciones: {usage.total_operations}")
    print(f"   Patrón: {usage.access_pattern}")
    
    if usage.complexity_impact.get('dominant_complexity'):
        print(f"   Complejidad dominante: {usage.complexity_impact['dominant_complexity']}")
    
    if usage.recommendations:
        print(f"   Recomendación: {usage.recommendations[0]}")
````

### 7.2 Análisis Comparativo

````
print("\nANÁLISIS DE COMPLEJIDAD")
print("\nImplementación original:")
print("  • Estructura: Array")
print("  • Búsquedas: O(n) cada una")
print("  • Total: O(n * m)")
print("\nImplementación optimizada (con hash table):")
print("  • Estructura: Hash Table")
print("  • Construcción: O(n)")
print("  • Búsquedas: O(1) cada una")
print("  • Total: O(n + m)")
print("\nMejora: De O(n*m) a O(n+m)")
````

## 8. Ejercicio Integrador

Analiza el siguiente código y genera recomendaciones:

````
code_exercise = """
algorithm socialNetwork(users[n], connections[m])
begin
    friends ← createArray(n)
    
    for i ← 1 to n do
        friends[i] ← createArray(0)
    
    for i ← 1 to m do
    begin
        u ← connections[i].from
        v ← connections[i].to
        
        call append(friends[u], v)
        call append(friends[v], u)
    end
    
    for i ← 1 to n do
    begin
        print("User " + i + " has " + length(friends[i]) + " friends")
        
        for friend in friends[i] do
            print("  - Friend: " + friend)
    end
end
"""

ast = parse_pseudocode(code_exercise)
result = identify_structures(ast, min_confidence=0.25)

print("EJERCICIO: RED SOCIAL")

print(f"\nEstructuras detectadas: {result.structure_count}")

for struct in result.structures_found:
    print(f"{struct.structure_name}")
    print(f"   Confianza: {struct.confidence:.2%}")
    
    usage = analyze_structure_usage(ast, struct)
    
    print(f"   Total operaciones: {usage.total_operations}")
    print(f"   Patrón: {usage.access_pattern}")
    
    if usage.recommendations:
        print(f"\n   Recomendaciones:")
        for rec in usage.recommendations:
            print(f"      • {rec}")
````

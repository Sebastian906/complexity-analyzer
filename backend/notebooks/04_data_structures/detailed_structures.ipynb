## 2. Arrays y Listas

### 2.1 Características Clave

**Arrays** son estructuras de acceso aleatorio con tamaño fijo o dinámico.

**Indicadores principales:**
- Parámetros con `A[n]` o `A[]`
- Accesos indexados `A[i]`, `A[j]`
- Iteraciones `for i ← 1 to n`

### 2.2 Ejemplo: Búsqueda Lineal

```
code_linear_search = """
algorithm linearSearch(A[n], x)
begin
    for i ← 1 to n do
    begin
        if A[i] = x then
            return i
    end
    return -1
end
"""

ast = parse_pseudocode(code_linear_search)
result = identify_structures(ast)
primary = result.primary_structure

print("BÚSQUEDA LINEAL")
print(f"Estructura: {primary.structure_name}")
print(f"Confianza: {primary.confidence:.2%}")
print(f"\nPropiedades:")
for k, v in primary.properties.items():
    print(f"  • {k}: {v}")
```

### 2.3 Ejemplo: Ordenamiento por Inserción

```
code_insertion_sort = """
algorithm insertionSort(A[n])
begin
    for i ← 2 to n do
    begin
        key ← A[i]
        j ← i - 1
        
        while j > 0 and A[j] > key do
        begin
            A[j+1] ← A[j]
            j ← j - 1
        end
        
        A[j+1] ← key
    end
end
"""

ast = parse_pseudocode(code_insertion_sort)
result = identify_structures(ast)

print("\nINSERTION SORT")
print(f"Total operaciones sobre array: {len(result.primary_structure.operations)}")
print("\nOperaciones:")
for op in result.primary_structure.operations[:5]:
    print(f"  • {op}")
```

### 2.4 Complejidades de Arrays

```
# Mostrar tabla de complejidades
complexities = []
for op in primary.operation_complexities:
    complexities.append({
        "Operación": op.operation,
        "Mejor": op.best_case,
        "Promedio": op.average_case,
        "Peor": op.worst_case
    })

df = pd.DataFrame(complexities)
print("\nCOMPLEJIDADES DE ARRAYS:")
print(df.to_string(index=False))
```

## 3. Pilas (Stacks)

### 3.1 Características LIFO

**Pilas** siguen el principio LIFO (Last In, First Out).

**Indicadores principales:**
- Operaciones `push()`, `pop()`
- Variables `stack`, `pila`
- Recursión (pila implícita)

### 3.2 Ejemplo: Pila Explícita

```
code_stack_explicit = """
algorithm balancedParentheses(expression)
begin
    stack ← createStack()
    
    for char in expression do
    begin
        if char = '(' or char = '[' or char = '{' then
            call push(stack, char)
        else if char = ')' or char = ']' or char = '}' then
        begin
            if isEmpty(stack) then
                return false
            
            top ← pop(stack)
            
            if not matches(top, char) then
                return false
        end
    end
    
    return isEmpty(stack)
end
"""

ast = parse_pseudocode(code_stack_explicit)
result = identify_structures(ast)
stack = result.primary_structure

print("PILA EXPLÍCITA - Paréntesis Balanceados")
print(f"Confianza: {stack.confidence:.2%}")
print("\nIndicadores encontrados:")
for ind in stack.indicators_found:
    print(f"  • {ind.name}: {ind.evidence}")
```

### 3.3 Ejemplo: Pila Implícita (Recursión)

```
code_factorial = """
algorithm factorial(n)
begin
    if n <= 1 then
        return 1
    
    return n * factorial(n - 1)
end
"""

ast = parse_pseudocode(code_factorial)
result = identify_structures(ast)

print("\nPILA IMPLÍCITA - Factorial Recursivo")

if result.primary_structure:
    stack = result.primary_structure
    print(f"Estructura: {stack.structure_name}")
    print(f"Confianza: {stack.confidence:.2%}")
    
    # Verificar indicador de recursión
    recursion_ind = next(
        (ind for ind in stack.indicators_found if "recursivo" in ind.name.lower()),
        None
    )
    if recursion_ind:
        print(f"\nPila implícita detectada: {recursion_ind.evidence}")
```

## 4. Colas (Queues)

### 4.1 Características FIFO

**Colas** siguen el principio FIFO (First In, First Out).

**Indicadores principales:**
- Operaciones `enqueue()`, `dequeue()`
- Variables `queue`, `cola`
- Uso en BFS

### 4.2 Ejemplo: BFS con Cola

```
code_bfs = """
algorithm bfs(graph, start)
begin
    queue ← createQueue()
    visited ← createSet()
    
    call enqueue(queue, start)
    call add(visited, start)
    
    while not isEmpty(queue) do
    begin
        node ← dequeue(queue)
        process(node)
        
        for neighbor in adjacency[node] do
        begin
            if neighbor not in visited then
            begin
                call enqueue(queue, neighbor)
                call add(visited, neighbor)
            end
        end
    end
end
"""

ast = parse_pseudocode(code_bfs)
result = identify_structures(ast, min_confidence=0.25)

print("BFS CON COLA")
print(f"Estructuras detectadas: {result.structure_count}")

for struct in result.structures_found:
    print(f"\n{struct.structure_name}")
    print(f"   Confianza: {struct.confidence:.2%}")
    print(f"   Variables: {', '.join(struct.variables)}")
```

## 5. Listas Enlazadas

### 5.1 Características

**Listas Enlazadas** usan nodos con punteros next/prev.

**Indicadores principales:**
- Acceso a `.next`, `.prev`
- Clase `Node`
- Travesía `while node != null`

### 5.2 Ejemplo: Travesía

```
code_linked_list = """
algorithm printList(head)
begin
    current ← head
    
    while current != null do
    begin
        print(current.data)
        current ← current.next
    end
end
"""

ast = parse_pseudocode(code_linked_list)
result = identify_structures(ast)

print("LISTA ENLAZADA - Travesía")

if result.primary_structure:
    ll = result.primary_structure
    print(f"Estructura: {ll.structure_name}")
    print(f"Confianza: {ll.confidence:.2%}")
    
    # Propiedades específicas
    if 'next_accesses' in ll.properties:
        print(f"\nAccesos a .next: {ll.properties['next_accesses']}")
```

## 6. Árboles

### 6.1 Características

**Árboles** son estructuras jerárquicas con raíz y nodos.

**Indicadores principales:**
- Acceso a `.left`, `.right`
- Recursión binaria (2 llamadas)
- Variables `tree`, `root`

### 6.2 Ejemplo: Recorrido Inorden

```
code_inorder = """
algorithm inorder(root)
begin
    if root != null then
    begin
        call inorder(root.left)
        process(root.data)
        call inorder(root.right)
    end
end
"""

ast = parse_pseudocode(code_inorder)
result = identify_structures(ast)

print("ÁRBOL - Recorrido Inorden")

if result.primary_structure:
    tree = result.primary_structure
    print(f"Confianza: {tree.confidence:.2%}")
    print("\nIndicadores:")
    for ind in tree.indicators_found:
        print(f"  • {ind.name}")
```

### 6.3 Ejemplo: Búsqueda en BST

```
code_bst_search = """
algorithm searchBST(root, key)
begin
    if root = null or root.data = key then
        return root
    
    if key < root.data then
        return searchBST(root.left, key)
    else
        return searchBST(root.right, key)
end
"""

ast = parse_pseudocode(code_bst_search)
result = identify_structures(ast)

print("\nBST - Búsqueda")
print(f"Estructura principal: {result.primary_structure.structure_name}")
print(f"Propiedades:")
for k, v in result.primary_structure.properties.items():
    print(f"  • {k}: {v}")
```

## 7. Grafos

### 7.1 Características

**Grafos** tienen nodos (vértices) y aristas.

**Indicadores principales:**
- Lista/matriz de adyacencia
- Variables `graph`, `adj`
- Array `visited[]`

### 7.2 Ejemplo: DFS

```
code_dfs = """
algorithm dfs(graph, start, n)
begin
    visited ← createArray(n)
    
    for i ← 1 to n do
        visited[i] ← false
    
    call dfsUtil(graph, start, visited)
end

algorithm dfsUtil(graph, node, visited)
begin
    visited[node] ← true
    process(node)
    
    for neighbor in graph[node] do
    begin
        if not visited[neighbor] then
            call dfsUtil(graph, neighbor, visited)
    end
end
"""

ast = parse_pseudocode(code_dfs)
result = identify_structures(ast, min_confidence=0.25)

print("GRAFO - DFS")

for struct in result.structures_found:
    print(f"\n{struct.structure_name}:")
    print(f"  Confianza: {struct.confidence:.2%}")
    if struct.code_evidence:
        print(f"  Evidencia: {struct.code_evidence[0]}")
```

## 9. Ejercicio Integrador

Analiza el siguiente código que usa múltiples estructuras:

```
code_dijkstra = """
algorithm dijkstra(graph, source, n)
begin
    dist ← createArray(n)
    visited ← createArray(n)
    pq ← createMinHeap()
    
    for i ← 1 to n do
    begin
        dist[i] ← infinity
        visited[i] ← false
    end
    
    dist[source] ← 0
    call insert(pq, source, 0)
    
    while not isEmpty(pq) do
    begin
        u ← extractMin(pq)
        
        if visited[u] then
            continue
        
        visited[u] ← true
        
        for neighbor in graph[u] do
        begin
            v ← neighbor.vertex
            weight ← neighbor.weight
            
            if dist[u] + weight < dist[v] then
            begin
                dist[v] ← dist[u] + weight
                call insert(pq, v, dist[v])
            end
        end
    end
    
    return dist
end
"""

ast = parse_pseudocode(code_dijkstra)
result = identify_structures(ast, min_confidence=0.2)

print("ALGORITMO DIJKSTRA")
print(f"Total estructuras: {result.structure_count}\n")

# Crear tabla resumen
summary_data = []
for struct in result.structures_found:
    summary_data.append({
        "Estructura": struct.structure_name,
        "Confianza": f"{struct.confidence:.2%}",
        "Variables": ", ".join(struct.variables[:2])
    })

df_summary = pd.DataFrame(summary_data)
print(df_summary.to_string(index=False))

print(f"\nEstructura principal: {result.primary_structure.structure_name}")
```

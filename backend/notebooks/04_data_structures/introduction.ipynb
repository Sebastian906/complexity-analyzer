## 2. Introducción al Módulo

El **Módulo de Detección de Estructuras de Datos** identifica automáticamente qué estructuras están siendo utilizadas en un algoritmo.

### ¿Qué detecta?

| Estructura | Descripción | Complejidad Típica |
|------------|-------------|-------------------|
| **Array/Lista** | Acceso secuencial/aleatorio | O(1) acceso |
| **Stack** | LIFO - Last In First Out | O(1) push/pop |
| **Queue** | FIFO - First In First Out | O(1) enqueue/dequeue |
| **Linked List** | Nodos con punteros | O(n) búsqueda |
| **Dictionary** | Clave-valor | O(1) promedio |
| **Tree** | Jerárquica | O(log n) - O(n) |
| **Graph** | Nodos y aristas | O(V+E) travesía |
| **Hash Table** | Hash + buckets | O(1) promedio |

### ¿Cómo funciona?

```
Pseudocódigo → Parser → AST → Detectores → Resultados
                                    ↓
                         [ArrayDetector, StackDetector, ...]
```

Cada detector:
1. Busca **indicadores** específicos en el AST
2. Calcula un **score de confianza** (0.0 - 1.0)
3. Identifica **operaciones** realizadas
4. Proporciona **complejidades** de cada operación

## 3. Ejemplo Básico

### Ejemplo 1: Detectar Array

````
# Código con array
code_array = """
algorithm bubbleSort(A[n])
begin
    for i ← 1 to n-1 do
        for j ← 1 to n-i do
            if A[j] > A[j+1] then
                temp ← A[j]
                A[j] ← A[j+1]
                A[j+1] ← temp
end
"""

# Parsear
ast = parse_pseudocode(code_array)

# Identificar estructuras
result = identify_structures(ast, min_confidence=0.3)

# Mostrar resultado
print("RESULTADO DE DETECCIÓN")
print(f"\n{result.summary}\n")

if result.primary_structure:
    primary = result.primary_structure
    print(f"Estructura: {primary.structure_name}")
    print(f"Confianza: {primary.confidence:.2%} ({primary.confidence_level.value})")
    print(f"Variables: {', '.join(primary.variables)}")
    print(f"\nRazonamiento:")
    print(f"   {primary.reasoning}")
````

**Salida esperada:**
```
============================================================
RESULTADO DE DETECCIÓN
============================================================

Estructura principal: Array/Lista (confianza: 85%). Total: 1 estructura(s) en 1 variable(s)

Estructura: Array/Lista
Confianza: 85% (high)
Variables: A

Razonamiento:
   Se encontraron 3 indicadores clave: Parámetro con corchetes, Acceso indexado, Iteración secuencial
```

### Ejemplo 2: Detectar Stack

````
# Código con pila
code_stack = """
algorithm reverseString(text)
begin
    stack ← createStack()
    
    for char in text do
        call push(stack, char)
    
    result ← ""
    while not isEmpty(stack) do
        char ← pop(stack)
        result ← result + char
    
    return result
end
"""

ast = parse_pseudocode(code_stack)
result = identify_structures(ast)

if result.primary_structure:
    primary = result.primary_structure
    print(f"Estructura: {primary.structure_name}")
    print(f"Confianza: {primary.confidence:.2%}")
    print(f"\nOperaciones detectadas:")
    for op in primary.operations[:5]:
        print(f"   • {op}")
````

## 4. Explorando Resultados

### 4.1 Indicadores Encontrados

````
# Usar código del ejemplo anterior
result = identify_structures(ast)
primary = result.primary_structure

print("INDICADORES ENCONTRADOS:")
for ind in primary.indicators_found:
    print(f"• {ind.name}")
    print(f"  Peso: {ind.weight:.1f} | Evidencia: {ind.evidence}")
    print()
````

### 4.2 Indicadores Faltantes

````
print("INDICADORES FALTANTES:")
for ind in primary.indicators_missing:
    print(f"• {ind.name} (peso: {ind.weight:.1f})")
````

### 4.3 Complejidades de Operaciones

````
print("COMPLEJIDADES:")

# Crear DataFrame para mejor visualización
complexities_data = []
for op_comp in primary.operation_complexities:
    complexities_data.append({
        "Operación": op_comp.operation,
        "Mejor Caso": op_comp.best_case,
        "Caso Promedio": op_comp.average_case,
        "Peor Caso": op_comp.worst_case,
        "Espacio": op_comp.space
    })

df = pd.DataFrame(complexities_data)
print(df.to_string(index=False))
````

### 4.4 Propiedades Específicas

````
print("PROPIEDADES:")
for key, value in primary.properties.items():
    print(f"• {key}: {value}")
````

### 4.5 Evidencia del Código

````
print("EVIDENCIA DEL CÓDIGO:")
for evidence in primary.code_evidence:
    print(f"• {evidence}")
````

## 5. Ejercicios

### Ejercicio 1: Múltiples Estructuras

Analiza el siguiente código que usa múltiples estructuras:

````
code_multiple = """
algorithm bfsWithStack(graph, start)
begin
    queue ← createQueue()
    visited ← createSet()
    
    call enqueue(queue, start)
    
    while not isEmpty(queue) do
    begin
        node ← dequeue(queue)
        
        if node not in visited then
        begin
            call add(visited, node)
            
            for neighbor in graph[node] do
                call enqueue(queue, neighbor)
        end
    end
end
"""

# Tu código aquí
ast = parse_pseudocode(code_multiple)
result = identify_structures(ast, min_confidence=0.25)

print(f"Total estructuras detectadas: {result.structure_count}")
print("\nEstructuras:")
for struct in result.structures_found:
    print(f"• {struct.structure_name}: {struct.confidence:.2%}")
````

**Preguntas:**
1. ¿Cuántas estructuras se detectaron?
2. ¿Cuál es la estructura principal?
3. ¿Por qué esa tiene mayor confianza?

## Tips

- **Umbral de confianza:** Usa 0.3-0.4 para exploración, 0.6+ para resultados confiables
- **Múltiples estructuras:** Es común que un algoritmo use varias estructuras
- **Pila implícita:** Los algoritmos recursivos usan pilas implícitas
- **Nombres de variables:** Ayudan mucho a la detección (usa `stack`, `queue`, etc.)

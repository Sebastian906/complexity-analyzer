## 2. ¿Qué es la caché?

La caché permite almacenar resultados de análisis, validación y exportación para acelerar operaciones repetitivas y reducir el tiempo de respuesta. Utiliza Redis por defecto, con fallback a caché local.

---

## 3. Servicios involucrados

- **CacheService:** Operaciones de caché (set, get, invalidate).
- **AnalysisOrchestrator:** Consulta la caché antes de ejecutar análisis.
- **ExportService:** Puede almacenar exportaciones generadas.

---

## 4. Almacenamiento y recuperación

```
# Guardar resultado en caché
cache.set("analysis_result:alg_123", analysis_result)

# Recuperar resultado de caché
cached_result = cache.get("analysis_result:alg_123")
```

- Es recomendable definir claves de caché claras y específicas para cada tipo de resultado.

---

## 5. Invalidación y actualización

```
# Invalidar caché
cache.invalidate("analysis_result:alg_123")
```

- Es importante invalidar la caché cuando se actualiza un algoritmo o sus resultados.

## 6. Integración avanzada

- El análisis completo puede consultar la caché antes de ejecutar operaciones costosas.
- Las exportaciones pueden almacenarse para descargas rápidas.
- Se puede usar la caché para almacenar resultados de validación y visualización.

```
# Ejemplo: uso en AnalysisOrchestrator
from app.services.analysis_orchestrator import AnalysisOrchestrator
orchestrator = AnalysisOrchestrator()

# Antes de analizar, buscar en caché
cached = cache.get(f"analysis_result:{algorithm_id}")
if cached is not None:
		print("Resultado recuperado de caché")
else:
		result = orchestrator.run_complete_analysis(algorithm_id)
		cache.set(f"analysis_result:{algorithm_id}", result)
```

---

## 7. Errores comunes y tips

- **Error:** No invalidar la caché tras actualizar un algoritmo.  
	**Solución:** Siempre invalidar la clave correspondiente.
- **Error:** Claves poco descriptivas.  
	**Solución:** Usa prefijos y nombres claros ("analysis_result:alg_123").
- **Tip:** Usa la caché para acelerar pruebas y desarrollo.
- **Tip:** Monitorea el tamaño de la caché para evitar sobrecarga.

## 2. ¿Qué es el análisis completo?

Es el proceso de analizar un algoritmo desde su registro hasta la obtención de métricas de complejidad, patrones, visualizaciones y reportes exportables.

---

## 3. Orquestación de servicios

- **AlgorithmService:** Recupera el algoritmo.
- **AnalysisOrchestrator:** Coordina el análisis completo.
- **PatternDetector:** Detecta patrones algorítmicos.
- **ExportService:** Exporta resultados.

## 6. Errores comunes y tips

- **Error:** No registrar el algoritmo antes de analizar.  
  **Solución:** Usa AlgorithmService para registrar y obtener el ID.
- **Error:** No validar el pseudocódigo antes de analizar.  
  **Solución:** Valida con el parser antes de ejecutar el análisis.
- **Tip:** Personaliza el flujo de análisis agregando validación con LLM o pasos extra.
- **Tip:** Exporta los resultados en varios formatos para documentación y reportes.

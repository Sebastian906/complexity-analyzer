## 2. ¿Qué es la validación?

La validación de algoritmos consiste en verificar que el pseudocódigo cumple con reglas sintácticas, semánticas y de buenas prácticas. Además, se puede usar un modelo de lenguaje (LLM) para sugerir mejoras o detectar errores lógicos.

---

## 3. Servicios y arquitectura

- **ValidationService:** Ejecuta validaciones automáticas y con LLMs.
- **ExportService:** Permite exportar los reportes de validación.
- **Parser:** Provee la estructura del algoritmo para validación.

---

## 4. Validación sintáctica y semántica

```
# Ejemplo de pseudocódigo
pseudocode = """
PROCEDIMIENTO Suma(a, b)
    c <- a + b
    RETORNAR c
FIN PROCEDIMIENTO
"""

# Parsear pseudocódigo
ast = parse_pseudocode(pseudocode)

# Validar sintaxis y semántica
service = ValidationService()
validation_result = service.validate_algorithm_from_ast(ast)

print(validation_result['rule_checks'])
```

---

## 5. Validación con LLMs

```
# Validar usando LLM (por ejemplo, Claude o Gemini)
llm_result = service.validate_with_llm(pseudocode)
print(llm_result['llm_feedback'])
```

---

## 6. Exportación de resultados

```
# Exportar reporte de validación
exporter = ExportService()
exporter.export_to_markdown(validation_result, output_path="./resultados/validacion.md")
```

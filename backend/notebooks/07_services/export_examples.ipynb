## 2. ¿Qué se puede exportar?

- Resultados de análisis de complejidad
- Reportes de validación
- Visualizaciones y diagramas

---

## 3. Formatos soportados

- PDF, JSON, Markdown, Excel, HTML, CSV, DOT, Mermaid, SVG

---

## 4. Exportación de análisis

```
# Exportar resultados de análisis a PDF
exporter.export_to_pdf(analysis_result, output_path="./resultados/analisis.pdf")

# Exportar a JSON
exporter.export_to_json(analysis_result, output_path="./resultados/analisis.json")
```

---

## 5. Exportación de validación

```
# Exportar reporte de validación a Markdown
exporter.export_to_markdown(validation_result, output_path="./resultados/validacion.md")
```

---

## 6. Exportación de visualizaciones

```
# Exportar visualización a SVG
exporter.export_to_svg(diagram_data, output_path="./resultados/diagrama.svg")

# Exportar a otros formatos
exporter.export_to_dot(diagram_data, output_path="./resultados/diagrama.dot")
```

## 7. Personalización de plantillas

---

## 8. Errores comunes y tips

- **Error:** Exportar un resultado vacío o sin analizar.  
	**Solución:** Asegúrate de ejecutar el análisis antes de exportar.
- **Error:** Sobrescribir archivos existentes sin advertencia.  
	**Solución:** Usa nombres de archivo únicos o verifica antes de exportar.
- **Tip:** Personaliza las plantillas para reportes profesionales.
- **Tip:** Exporta en varios formatos para compartir con diferentes audiencias.

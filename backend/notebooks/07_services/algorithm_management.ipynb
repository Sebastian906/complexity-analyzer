## 4. Registro y actualización

```
# Registrar un algoritmo
service.register_algorithm(name="Bubble Sort", code="""
PROCEDIMIENTO BubbleSort(A, n)
		PARA i <- 1 HASTA n-1 HACER
				PARA j <- 1 HASTA n-i HACER
						SI A[j] > A[j+1] ENTONCES
								temp <- A[j]
								A[j] <- A[j+1]
								A[j+1] <- temp
						FIN SI
				FIN PARA
		FIN PARA
FIN PROCEDIMIENTO
""")

# Actualizar un algoritmo
service.update_algorithm(algorithm_id, new_code="...nuevo código...")
```

## 8. Errores comunes y tips

- **Error:** Registrar pseudocódigo inválido.  
	**Solución:** Validar con el parser antes de guardar.
- **Error:** Duplicar nombres de algoritmos.  
	**Solución:** Usar nombres únicos o asociar metadatos.
- **Tip:** Implementa control de versiones para auditoría y trazabilidad.
- **Tip:** Usa tags para organizar y buscar algoritmos fácilmente.

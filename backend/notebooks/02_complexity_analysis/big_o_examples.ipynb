## Introduccion a Big O

La notacion Big O describe el comportamiento asintotico de un algoritmo en el **peor caso**.

### Definicion Formal

Decimos que f(n) = O(g(n)) si existen constantes positivas c y n0 tales que:

```
f(n) <= c * g(n) para todo n >= n0
```

### Complejidades Comunes (ordenadas)

| Notacion | Nombre | Ejemplo |
|----------|--------|----------|
| O(1) | Constante | Acceso a array |
| O(log n) | Logaritmica | Busqueda binaria |
| O(sqrt(n)) | Raiz cuadrada | Algoritmo de factorizacion |
| O(n) | Lineal | Busqueda lineal |
| O(n log n) | Linealitmica | Merge Sort, Quick Sort |
| O(n^2) | Cuadratica | Bubble Sort, Insertion Sort |
| O(n^3) | Cubica | Multiplicacion de matrices |
| O(2^n) | Exponencial | Subconjuntos, Fibonacci recursivo |
| O(n!) | Factorial | Permutaciones |

## Analisis de Estructuras Basicas

### Sentencias Simples: O(1)

Las siguientes operaciones son O(1):

- Asignacion: `x <- 5`
- Operaciones aritmeticas: `x <- a + b * c`
- Comparaciones: `if (x > y)`
- Acceso a array: `A[i]`
- Retorno: `return x`

### Ciclo For Simple: O(n)

```
for i <- 1 to n do
    x <- x + 1    // O(1)
end
```

Analisis:
- El ciclo itera n veces
- Cada iteracion es O(1)
- Total: O(n * 1) = O(n)

### Ciclos Anidados: O(n^2)

```
for i <- 1 to n do
    for j <- 1 to n do
        x <- x + 1    // O(1)
    end
end
```

Analisis:
- Ciclo externo: n iteraciones
- Ciclo interno: n iteraciones por cada externa
- Total: O(n * n) = O(n^2)

### Ciclo con Limite Variable: O(n^2)

```
for i <- 1 to n do
    for j <- 1 to i do
        x <- x + 1
    end
end
```

Analisis detallado:
- i=1: j itera 1 vez
- i=2: j itera 2 veces
- i=3: j itera 3 veces
- ...
- i=n: j itera n veces

Total de iteraciones: 1 + 2 + 3 + ... + n = n(n+1)/2 = O(n^2)

### Ciclo Logaritmico: O(log n)

```
i <- n
while (i >= 1) do
    x <- x + 1
    i <- i / 2
end
```

Analisis:
- En cada iteracion, i se reduce a la mitad
- Numero de iteraciones: log2(n)
- Total: O(log n)

## Ejemplos por Complejidad

### O(1) - Constante

```
algorithm getFirst(A[], n)
begin
    return A[1]
end
```

El tiempo no depende del tamano de entrada.

### O(log n) - Busqueda Binaria

```
algorithm binarySearch(A[], n, key)
begin
    low <- 1
    high <- n
    while (low <= high) do
        mid <- floor((low + high) / 2)
        if (A[mid] = key) then
            return mid
        else
            if (A[mid] < key) then
                low <- mid + 1
            else
                high <- mid - 1
            end
        end
    end
    return -1
end
```

En cada iteracion se descarta la mitad del espacio de busqueda.

### O(n) - Busqueda Lineal

```
algorithm linearSearch(A[], n, key)
begin
    for i <- 1 to n do
        if (A[i] = key) then
            return i
        end
    end
    return -1
end
```

En el peor caso, se recorren todos los n elementos.

### O(n log n) - Merge Sort

```
algorithm mergeSort(A[], p, r)
begin
    if (p < r) then
        q <- floor((p + r) / 2)
        call mergeSort(A, p, q)          // T(n/2)
        call mergeSort(A, q + 1, r)      // T(n/2)
        call merge(A, p, q, r)           // O(n)
    end
end
```

Ecuacion de recurrencia: T(n) = 2T(n/2) + O(n)

Por Teorema Maestro: O(n log n)

### O(n^2) - Bubble Sort

```
algorithm bubbleSort(A[], n)
begin
    for i <- 1 to n - 1 do
        for j <- 1 to n - i do
            if (A[j] > A[j + 1]) then
                temp <- A[j]
                A[j] <- A[j + 1]
                A[j + 1] <- temp
            end
        end
    end
end
```

Dos ciclos anidados que recorren el array.

### O(2^n) - Fibonacci Recursivo

```
algorithm fibonacci(n)
begin
    if (n <= 1) then
        return n
    else
        return call fibonacci(n - 1) + call fibonacci(n - 2)
    end
end
```

Cada llamada genera dos nuevas llamadas, formando un arbol binario.

## Reglas de Combinacion

### Regla de la Suma (Secuencial)

Si dos bloques se ejecutan en secuencia:

```
O(f(n)) + O(g(n)) = O(max(f(n), g(n)))
```

Ejemplo:

```
// Bloque 1: O(n)
for i <- 1 to n do
    x <- x + 1
end

// Bloque 2: O(n^2)
for i <- 1 to n do
    for j <- 1 to n do
        y <- y + 1
    end
end
```

Total: O(n) + O(n^2) = O(n^2)

### Regla del Producto (Anidado)

Si un bloque esta dentro de otro:

```
O(f(n)) * O(g(n)) = O(f(n) * g(n))
```

Ejemplo:

```
for i <- 1 to n do           // O(n)
    for j <- 1 to n do       // O(n)
        x <- x + 1           // O(1)
    end
end
```

Total: O(n) * O(n) * O(1) = O(n^2)

### Regla del Condicional (If)

En analisis Big O (peor caso), se toma el maximo:

```
if (condicion) then
    // Bloque A: O(n)
else
    // Bloque B: O(n^2)
end
```

Total: max(O(n), O(n^2)) = O(n^2)

## Implementacion en el Sistema

### Metodo analyze_node()

El analizador procesa cada tipo de nodo del AST:

| Tipo de Nodo | Metodo | Logica |
|--------------|--------|--------|
| `BlockNode` | `analyze_block()` | Combina secuencialmente |
| `ForLoopNode` | `_analyze_for_loop()` | Iteraciones * cuerpo |
| `WhileLoopNode` | `_analyze_while_loop()` | Asume O(n) |
| `RepeatLoopNode` | `_analyze_repeat_loop()` | Similar a while |
| `IfStatementNode` | `_analyze_if_statement()` | Max de ramas |
| `AssignmentNode` | - | Retorna "1" |
| `CallStatementNode` | - | Retorna "1" (por defecto) |

## 2. Notacion Big O

La notacion Big O describe el comportamiento asintotico de un algoritmo en el **peor caso**.

### Definicion Formal

Decimos que f(n) = O(g(n)) si existen constantes positivas c y n0 tales que:

```
f(n) <= c * g(n) para todo n >= n0
```

### Complejidades Comunes (ordenadas)

| Notacion | Nombre | Ejemplo |
|----------|--------|----------|
| O(1) | Constante | Acceso a array |
| O(log n) | Logaritmica | Busqueda binaria |
| O(n) | Lineal | Busqueda lineal |
| O(n log n) | Linealitmica | Merge Sort |
| O(n^2) | Cuadratica | Bubble Sort |
| O(2^n) | Exponencial | Fibonacci recursivo |
| O(n!) | Factorial | Permutaciones |

## 4. Ejemplos Practicos

### Ejemplo 1: O(1) - Constante

````python
# O(1) - Acceso directo
codigo_o1 = """
algorithm getFirst(A[], n)
begin
    return A[1]
end
"""

ast = parse_pseudocode(codigo_o1)
analyzer = BigOAnalyzer()
resultado = analyzer.analyze(ast)

print("ANALISIS O(1):")
print(f"Algoritmo: {ast.algorithm.name}")
print(f"Complejidad: {resultado}")
````

### Ejemplo 2: O(n) - Lineal

````python
# O(n) - Busqueda lineal
codigo_on = """
algorithm linearSearch(A[], n, key)
begin
    for i <- 1 to n do
        if (A[i] = key) then
            return i
        end
    end
    return -1
end
"""

ast = parse_pseudocode(codigo_on)
resultado = analyzer.analyze(ast)

print("ANALISIS O(n):")
print(f"Algoritmo: {ast.algorithm.name}")
print(f"Complejidad: {resultado}")
````

### Ejemplo 3: O(n^2) - Cuadratica

````python
# O(n^2) - Bubble Sort
codigo_on2 = """
algorithm bubbleSort(A[], n)
begin
    for i <- 1 to n - 1 do
        for j <- 1 to n - i do
            if (A[j] > A[j + 1]) then
                temp <- A[j]
                A[j] <- A[j + 1]
                A[j + 1] <- temp
            end
        end
    end
end
"""

ast = parse_pseudocode(codigo_on2)
resultado = analyzer.analyze(ast)

print("ANALISIS O(n^2):")
print(f"Algoritmo: {ast.algorithm.name}")
print(f"Complejidad: {resultado}")
````

### Comparacion de Complejidades

````python
# Comparar varios algoritmos
algoritmos = [
    ("getFirst", codigo_o1),
    ("linearSearch", codigo_on),
    ("bubbleSort", codigo_on2)
]

resultados = []
for nombre, codigo in algoritmos:
    ast = parse_pseudocode(codigo)
    resultado = analyzer.analyze(ast)
    resultados.append({
        "Algoritmo": nombre,
        "Complejidad": str(resultado)
    })

df = pd.DataFrame(resultados)
print("COMPARACION DE COMPLEJIDADES:")
print(df.to_string(index=False))
````

## 5. Reglas de Combinacion

### Regla de la Suma (Secuencial)

```
O(f(n)) + O(g(n)) = O(max(f(n), g(n)))
```

### Regla del Producto (Anidado)

```
O(f(n)) * O(g(n)) = O(f(n) * g(n))
```

### Ejemplo Practico

````python
# Codigo con bloques secuenciales
codigo_secuencial = """
algorithm mixedComplexity(A[], n)
begin
    for i <- 1 to n do
        x <- x + 1
    end
    
    for i <- 1 to n do
        for j <- 1 to n do
            y <- y + 1
        end
    end
end
"""

ast = parse_pseudocode(codigo_secuencial)
resultado = analyzer.analyze(ast)

print("REGLA DE LA SUMA:")
print("Bloque 1: O(n)")
print("Bloque 2: O(n^2)")
print(f"Total: O(n) + O(n^2) = {resultado}")
````

### Usando ComplexityCalculator (Orquestador)

````python
# Usar ComplexityCalculator para analisis completo
calculator = ComplexityCalculator()
ast = parse_pseudocode(codigo_on2)

results = calculator.analyze_all(ast)

print("ANALISIS COMPLETO:")
print(f"Big O (peor caso):  {results['big_o']}")
print(f"Omega (mejor caso): {results['omega']}")
print(f"Theta (ajustada):   {results['theta']}")
````

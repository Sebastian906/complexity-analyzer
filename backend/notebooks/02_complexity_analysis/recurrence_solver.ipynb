## Introduccion a las Recurrencias

Una ecuacion de recurrencia define el tiempo de ejecucion de un algoritmo recursivo
en terminos de si mismo aplicado a entradas mas pequenas.

### Componentes de una Recurrencia

| Componente | Descripcion | Ejemplo |
|------------|-------------|----------|
| T(n) | Tiempo para entrada de tamano n | T(n) |
| Caso base | Condicion de parada | T(1) = O(1) |
| Llamadas recursivas | Subproblemas | T(n/2), T(n-1) |
| Trabajo fuera de recursion | Operaciones adicionales | O(n), O(1) |

### Ejemplo: Merge Sort

```
T(n) = 2T(n/2) + O(n)
T(1) = O(1)
```

Interpretacion:
- 2T(n/2): Dos llamadas recursivas con mitad del tamano
- O(n): Trabajo de merge (combinar dos mitades)

## Formas de Ecuaciones

El sistema reconoce 7 formas de ecuaciones de recurrencia:

### Divide y Venceras

| Forma | Ecuacion | Ejemplo |
|-------|----------|----------|
| F0 | T(n) = T(n/b) + f(n) | Busqueda binaria |
| F1 | T(n) = aT(n/b) + f(n) | Merge Sort, Quick Sort |
| F2 | T(n) = T(n/b) + T(n/c) + f(n) | Division no uniforme |
| F3 | T(n) = T(n/b) + T(n/c) + ... + f(n) | Multiples divisiones |

### Resta y Venceras

| Forma | Ecuacion | Ejemplo |
|-------|----------|----------|
| F4 | T(n) = T(n-b) + f(n) | Factorial |
| F5 | T(n) = aT(n-b) + f(n) | Fibonacci, Torres de Hanoi |
| F6 | T(n) = aT(n-b) + cT(n-d) + f(n) | Recurrencia mixta |

### Patrones de Recurrencia

```python
@dataclass
class RecurrencePattern:
    form: RecurrenceForm     # F0, F1, F2, ..., F6
    a: int = 1               # Numero de llamadas
    b: int = 2               # Factor de division/resta
    c: Optional[int] = None  # Segundo factor (F6)
    d: Optional[int] = None  # Segundo decremento (F6)
    f_n: str = "1"           # Trabajo fuera de recursion
```

## Metodos de Solucion

El sistema implementa 5 metodos para resolver ecuaciones:

### 1. Iteracion

Expande la recurrencia repetidamente hasta llegar al caso base.

Ejemplo para T(n) = T(n-1) + 1:

```
T(n) = T(n-1) + 1
     = T(n-2) + 1 + 1
     = T(n-3) + 1 + 1 + 1
     = ...
     = T(1) + (n-1)
     = 1 + n - 1
     = O(n)
```

### 2. Arbol de Recursion

Visualiza las llamadas como un arbol y suma el trabajo en cada nivel.

Ejemplo para T(n) = 2T(n/2) + n:

```
Nivel 0:                    n                   (trabajo: n)
Nivel 1:          n/2              n/2          (trabajo: n)
Nivel 2:      n/4    n/4      n/4    n/4        (trabajo: n)
...
Nivel k:   n/2^k nodos con trabajo n/2^k cada uno

Profundidad: log2(n)
Trabajo por nivel: n
Total: n * log(n) = O(n log n)
```

### 3. Teorema Maestro

Aplica directamente para recurrencias de la forma T(n) = aT(n/b) + f(n).

(Detallado en la siguiente seccion)

### 4. Sustitucion Inteligente

Adivina la solucion y la verifica por induccion.

### 5. Ecuacion Caracteristica

Para recurrencias lineales con coeficientes constantes.

Ejemplo para T(n) = T(n-1) + T(n-2):

```
Ecuacion caracteristica: x^2 = x + 1
Raices: phi = (1 + sqrt(5))/2, psi = (1 - sqrt(5))/2
Solucion: T(n) = c1*phi^n + c2*psi^n = O(phi^n) = O(1.618^n)
```

## Teorema Maestro

El Teorema Maestro resuelve recurrencias de la forma:

```
T(n) = aT(n/b) + f(n)
```

Donde:
- a >= 1: numero de subproblemas
- b > 1: factor de reduccion
- f(n): trabajo fuera de recursion

### Los Tres Casos

Sea c_crit = log_b(a) (exponente critico)

| Caso | Condicion | Solucion |
|------|-----------|----------|
| 1 | f(n) = O(n^(c_crit - epsilon)) | T(n) = Theta(n^c_crit) |
| 2 | f(n) = Theta(n^c_crit) | T(n) = Theta(n^c_crit * log n) |
| 3 | f(n) = Omega(n^(c_crit + epsilon)) | T(n) = Theta(f(n)) |

### Ejemplo: Merge Sort

T(n) = 2T(n/2) + n

- a = 2, b = 2, f(n) = n
- c_crit = log_2(2) = 1
- f(n) = n = n^1 = n^c_crit
- Caso 2: T(n) = Theta(n^1 * log n) = Theta(n log n)

### Ejemplo: Busqueda Binaria

T(n) = T(n/2) + 1

- a = 1, b = 2, f(n) = 1
- c_crit = log_2(1) = 0
- f(n) = 1 = n^0 = n^c_crit
- Caso 2: T(n) = Theta(n^0 * log n) = Theta(log n)

### Ejemplo: Strassen

T(n) = 7T(n/2) + n^2

- a = 7, b = 2, f(n) = n^2
- c_crit = log_2(7) aproximado 2.807
- f(n) = n^2 = O(n^(2.807 - 0.807))
- Caso 1: T(n) = Theta(n^log_2(7)) aproximado Theta(n^2.807)

## Tabla de Aplicabilidad

Cada metodo aplica a ciertas formas:

| Metodo | Formas Aplicables |
|--------|-------------------|
| Iteracion | F0, F1, F4, F5 |
| Arbol de Recursion | F0, F1, F2, F3, F5, F6 |
| Teorema Maestro | F0, F1 |
| Sustitucion Inteligente | Todas |
| Ecuacion Caracteristica | F4, F5, F6 |

### Seleccion Automatica de Metodo

El sistema selecciona el metodo mas apropiado:

1. Para F0/F1 con f(n) polinomial: Teorema Maestro
2. Para F4/F5/F6 lineales: Ecuacion Caracteristica
3. Como fallback: Arbol de Recursion o Iteracion

## Ejemplos de Resolucion

### Factorial: T(n) = T(n-1) + O(1)

Forma: F4 (Resta y Venceras)

Resolucion por Iteracion:

```
T(n) = T(n-1) + 1
     = T(n-2) + 2
     = T(n-3) + 3
     = ...
     = T(1) + (n-1)
     = O(n)
```

### Fibonacci: T(n) = T(n-1) + T(n-2) + O(1)

Forma: F6 (Resta y seras Vencido)

Resolucion por Ecuacion Caracteristica:

```
x^2 = x + 1
x^2 - x - 1 = 0
x = (1 +- sqrt(5)) / 2

phi = (1 + sqrt(5)) / 2 aproximado 1.618

T(n) = O(phi^n) = O(1.618^n)
```

### Quick Sort (caso promedio): T(n) = 2T(n/2) + O(n)

Forma: F1 (Divide y Venceras)

Resolucion por Teorema Maestro:

```
a = 2, b = 2, f(n) = n
c_crit = log_2(2) = 1
f(n) = n = Theta(n^1)
Caso 2: T(n) = Theta(n log n)
```

### Torres de Hanoi: T(n) = 2T(n-1) + O(1)

Forma: F5 (Resta y seras Vencido)

Resolucion por Arbol de Recursion:

```
Nivel 0: 1 nodo, trabajo 1
Nivel 1: 2 nodos, trabajo 2
Nivel 2: 4 nodos, trabajo 4
...
Nivel k: 2^k nodos

Profundidad: n
Total: 1 + 2 + 4 + ... + 2^(n-1) = 2^n - 1 = O(2^n)
```

## Estructura del Resultado

El resolvedor retorna un objeto `SolutionResult`:

```python
@dataclass
class SolutionResult:
    complexity: str              # "n log n", "2^n", etc.
    method_used: SolutionMethod  # MASTER_THEOREM, ITERATION, etc.
    steps: List[str]             # Pasos de la resolucion
    form_detected: RecurrenceForm # F0, F1, ..., F6
    is_exact: bool = True        # Si la solucion es exacta
    explanation: str = ""        # Explicacion textual
    alternative_methods: List[SolutionMethod] = []  # Otros metodos aplicables
```

### Ejemplo de Resultado

Para T(n) = 2T(n/2) + n:

```python
SolutionResult(
    complexity="n log n",
    method_used=SolutionMethod.MASTER_THEOREM,
    steps=[
        "Identificada forma F1: T(n) = aT(n/b) + f(n)",
        "Parametros: a=2, b=2, f(n)=n",
        "Calculando c_crit = log_2(2) = 1",
        "f(n) = n = Theta(n^1), coincide con n^c_crit",
        "Aplicando Caso 2 del Teorema Maestro",
        "Solucion: T(n) = Theta(n^1 * log n) = Theta(n log n)"
    ],
    form_detected=RecurrenceForm.F1,
    is_exact=True,
    explanation="El algoritmo divide en 2 subproblemas de tamaño n/2...",
    alternative_methods=[SolutionMethod.RECURSION_TREE, SolutionMethod.ITERATION]
)
```

## Conclusiones

El sistema de resolucion de recurrencias:

1. **Detecta automaticamente** la forma de la ecuacion (F0-F6)
2. **Selecciona el metodo optimo** basado en la forma
3. **Proporciona pasos detallados** de la resolucion
4. **Sugiere metodos alternativos** cuando estan disponibles

Esto permite analizar algoritmos recursivos como:
- Divide y Venceras (Merge Sort, Quick Sort, Strassen)
- Recursion lineal (Factorial, Fibonacci)
- Backtracking (N-Reinas, Sudoku)

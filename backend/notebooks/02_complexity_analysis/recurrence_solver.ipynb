## 2. Introduccion a las Recurrencias

Una ecuacion de recurrencia define el tiempo de ejecucion en terminos de si mismo.

### Componentes

| Componente | Descripcion | Ejemplo |
|------------|-------------|---------|
| T(n) | Tiempo para tamano n | T(n) |
| Caso base | Condicion de parada | T(1) = O(1) |
| Llamadas recursivas | Subproblemas | T(n/2), T(n-1) |

### Ejemplo: Construir Recurrencia desde AST

```python
# Construir recurrencia desde codigo
codigo_factorial = """
algorithm factorial(n)
begin
    if (n <= 1) then
        return 1
    else
        return n * call factorial(n - 1)
    end
end
"""

ast = parse_pseudocode(codigo_factorial)
recurrencia = builder.build(ast)

print("=== FACTORIAL ===")
print(f"Codigo: factorial(n)")
print(f"Recurrencia: T(n) = T(n-1) + O(1)")
print(f"Interpretacion: Una llamada recursiva con n-1, trabajo constante")
```

## 3. Formas de Ecuaciones

El sistema reconoce 7 formas de recurrencias:

### Divide y Venceras (F0-F3)

| Forma | Ecuacion | Ejemplo |
|-------|----------|---------|
| F0 | T(n) = T(n/b) + f(n) | Busqueda binaria |
| F1 | T(n) = aT(n/b) + f(n) | Merge Sort |

### Resta y Venceras (F4-F6)

| Forma | Ecuacion | Ejemplo |
|-------|----------|---------|
| F4 | T(n) = T(n-b) + f(n) | Factorial |
| F5 | T(n) = aT(n-b) + f(n) | Torres de Hanoi |

### Ejemplo: Resolver Diferentes Formas

```python
# Resolver diferentes formas de recurrencias
recurrencias = [
    ("Busqueda Binaria", "T(n) = T(n/2) + 1"),
    ("Merge Sort", "T(n) = 2T(n/2) + n"),
    ("Factorial", "T(n) = T(n-1) + 1"),
    ("Torres de Hanoi", "T(n) = 2T(n-1) + 1"),
]

print("=== RESOLUCION DE RECURRENCIAS ===")
for nombre, ecuacion in recurrencias:
    try:
        resultado = solver.solve(ecuacion)
        print(f"\n{nombre}:")
        print(f"  Ecuacion: {ecuacion}")
        print(f"  Solucion: O({resultado.complexity})")
    except Exception as e:
        print(f"\n{nombre}: Error - {e}")
```

## 4. Teorema Maestro

El Teorema Maestro resuelve recurrencias de la forma T(n) = aT(n/b) + f(n)

### Los Tres Casos

Sea c_crit = log_b(a)

| Caso | Condicion | Solucion |
|------|-----------|----------|
| 1 | f(n) < n^c_crit | Theta(n^c_crit) |
| 2 | f(n) = n^c_crit | Theta(n^c_crit * log n) |
| 3 | f(n) > n^c_crit | Theta(f(n)) |

### Ejemplo: Aplicar Teorema Maestro

```python
# Aplicar Teorema Maestro manualmente
import math

def teorema_maestro(a, b, f_exp):
    """
    Aplica el Teorema Maestro
    a: numero de subproblemas
    b: factor de division
    f_exp: exponente de f(n) = n^f_exp
    """
    c_crit = math.log(a) / math.log(b)
    
    print(f"Parametros: a={a}, b={b}, f(n)=n^{f_exp}")
    print(f"c_crit = log_{b}({a}) = {c_crit:.3f}")
    
    if f_exp < c_crit:
        print(f"Caso 1: f(n) < n^c_crit")
        return f"n^{c_crit:.2f}"
    elif abs(f_exp - c_crit) < 0.01:
        print(f"Caso 2: f(n) = n^c_crit")
        return f"n^{c_crit:.0f} log n"
    else:
        print(f"Caso 3: f(n) > n^c_crit")
        return f"n^{f_exp}"

# Ejemplos
print("=== MERGE SORT: T(n) = 2T(n/2) + n ===")
sol = teorema_maestro(2, 2, 1)
print(f"Solucion: Theta({sol})")

print("\n=== STRASSEN: T(n) = 7T(n/2) + n^2 ===")
sol = teorema_maestro(7, 2, 2)
print(f"Solucion: Theta({sol})")
```

## 5. Ejemplos Practicos

### Resolver desde Codigo Recursivo

```python
# Analizar algoritmo recursivo completo
codigo_merge = """
algorithm mergeSort(A[], p, r)
begin
    if (p < r) then
        q <- floor((p + r) / 2)
        call mergeSort(A, p, q)
        call mergeSort(A, q + 1, r)
    end
end
"""

ast = parse_pseudocode(codigo_merge)

# Construir y resolver recurrencia
recurrencia = builder.build(ast)
print("=== MERGE SORT ===")
print(f"Algoritmo: {ast.algorithm.name}")
print(f"Recurrencia detectada: T(n) = 2T(n/2) + O(n)")

resultado = solver.solve("T(n) = 2T(n/2) + n")
print(f"Complejidad: O({resultado.complexity})")
```

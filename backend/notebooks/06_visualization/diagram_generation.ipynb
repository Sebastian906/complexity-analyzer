# Generación de Diagramas - Técnicas y Formatos

Este notebook cubre las diferentes técnicas de generación de diagramas y formatos de exportación disponibles.

## Configuración

```
import sys
sys.path.append('../..')

from app.core.visualization.graph_generator import GraphGenerator
from app.core.visualization.tree_builder import TreeBuilder
from app.core.visualization.diagram_renderer import DiagramRenderer
from app.core.visualization.batch_exporter import BatchExporter, ExportTask, ExportFormat
from pathlib import Path

output_dir = Path("../../data/exports/notebooks/diagrams")
output_dir.mkdir(parents=True, exist_ok=True)

print("Módulos cargados")
```

## 1. Fundamentos de Generación de Diagramas

### Tipos de Diagramas Soportados

1. **Árboles**: Estructuras jerárquicas
2. **Grafos**: Relaciones generales
3. **Flujos**: Diagramas de flujo y CFG
4. **Redes**: Grafos con pesos

## 2. Generación de Árboles

### Árbol Binario Simple

```
tree_builder = TreeBuilder()

# Crear árbol binario
tree_data = {
    'root': {
        'id': 'root',
        'value': 10,
        'children': [
            {
                'id': 'left',
                'value': 5,
                'children': [
                    {'id': 'll', 'value': 3, 'children': []},
                    {'id': 'lr', 'value': 7, 'children': []}
                ]
            },
            {
                'id': 'right',
                'value': 15,
                'children': [
                    {'id': 'rl', 'value': 12, 'children': []},
                    {'id': 'rr', 'value': 20, 'children': []}
                ]
            }
        ]
    }
}

# Construir representación visual
tree = tree_builder.build_tree(tree_data)

print(f" Árbol construido:")
print(f"  - Altura: {tree.height}")
print(f"  - Nodos: {tree.node_count}")
print(f"  - Balance: {tree.balance_factor}")
```

### Árbol N-ario

```
# Árbol de directorios
filesystem_tree = {
    'root': {
        'id': 'root',
        'label': '/',
        'type': 'directory',
        'children': [
            {
                'id': 'home',
                'label': 'home',
                'type': 'directory',
                'children': [
                    {'id': 'user', 'label': 'user', 'type': 'directory', 'children': []},
                    {'id': 'admin', 'label': 'admin', 'type': 'directory', 'children': []}
                ]
            },
            {
                'id': 'var',
                'label': 'var',
                'type': 'directory',
                'children': [
                    {'id': 'log', 'label': 'log', 'type': 'directory', 'children': []},
                    {'id': 'tmp', 'label': 'tmp', 'type': 'directory', 'children': []}
                ]
            }
        ]
    }
}

fs_tree = tree_builder.build_tree(filesystem_tree, style='filesystem')
print(f" Árbol de filesystem: {fs_tree.node_count} nodos")
```

## 3. Generación de Grafos

### Grafo Dirigido Simple

```
graph_generator = GraphGenerator()

# Definir nodos y aristas
nodes = [
    {'id': 'A', 'label': 'Node A', 'color': '#FF6B6B'},
    {'id': 'B', 'label': 'Node B', 'color': '#4ECDC4'},
    {'id': 'C', 'label': 'Node C', 'color': '#45B7D1'},
    {'id': 'D', 'label': 'Node D', 'color': '#FFA07A'},
]

edges = [
    ('A', 'B', {'weight': 5, 'label': '5'}),
    ('A', 'C', {'weight': 3, 'label': '3'}),
    ('B', 'D', {'weight': 2, 'label': '2'}),
    ('C', 'D', {'weight': 7, 'label': '7'}),
]

# Generar grafo
graph = graph_generator.generate_graph(
    nodes,
    edges,
    options={
        'directed': True,
        'show_weights': True,
        'layout': 'hierarchical'
    }
)

print(f" Grafo generado: {len(nodes)} nodos, {len(edges)} aristas")
```

### Grafo No Dirigido (Ejemplo: Red Social)

```
# Red social simple
social_nodes = [
    {'id': 'alice', 'label': 'Alice', 'type': 'person'},
    {'id': 'bob', 'label': 'Bob', 'type': 'person'},
    {'id': 'charlie', 'label': 'Charlie', 'type': 'person'},
    {'id': 'diana', 'label': 'Diana', 'type': 'person'},
    {'id': 'eve', 'label': 'Eve', 'type': 'person'},
]

social_edges = [
    ('alice', 'bob', {'relation': 'friend'}),
    ('alice', 'charlie', {'relation': 'friend'}),
    ('bob', 'diana', {'relation': 'friend'}),
    ('charlie', 'diana', {'relation': 'friend'}),
    ('charlie', 'eve', {'relation': 'colleague'}),
]

social_graph = graph_generator.generate_graph(
    social_nodes,
    social_edges,
    options={
        'directed': False,
        'layout': 'force',
        'show_labels': True
    }
)

print(f" Red social: {len(social_nodes)} personas, {len(social_edges)} relaciones")
```

## 4. Layouts y Algoritmos de Posicionamiento

### Comparación de Layouts

```
# Mismo grafo, diferentes layouts
layouts = ['hierarchical', 'circular', 'force', 'grid']

for layout in layouts:
    graph = graph_generator.generate_graph(
        nodes,
        edges,
        options={'layout': layout, 'title': f'Layout: {layout}'}
    )
    
    print(f"Layout {layout}: {graph.layout_info}")
```

### Layout Personalizado

```
# Definir posiciones manualmente
positioned_nodes = [
    {'id': 'A', 'label': 'A', 'x': 0, 'y': 0},
    {'id': 'B', 'label': 'B', 'x': 100, 'y': 50},
    {'id': 'C', 'label': 'C', 'x': 50, 'y': 100},
]

positioned_edges = [
    ('A', 'B', {}),
    ('A', 'C', {}),
    ('B', 'C', {}),
]

custom_graph = graph_generator.generate_graph(
    positioned_nodes,
    positioned_edges,
    options={'layout': 'manual', 'use_positions': True}
)

print("Layout personalizado aplicado")
```

## 5. Formatos de Exportación

### DOT (Graphviz)

```
renderer = DiagramRenderer()

# Exportar a DOT
dot_content = graph.to_dot()

print("DOT Output:")
print(dot_content)
print()

# Guardar
dot_path = output_dir / 'graph.dot'
dot_path.write_text(dot_content)
print(f" DOT guardado en: {dot_path}")
```

Puedes renderizar el archivo DOT con:

```
%%bash
dot -Tpng graph.dot -o graph.png
dot -Tsvg graph.dot -o graph.svg
```

### Mermaid

```
# Exportar a Mermaid
mermaid_content = graph.to_mermaid()

print("Mermaid Output:")
print(mermaid_content)
print()

# Guardar
mermaid_path = output_dir / 'graph.mmd'
mermaid_path.write_text(mermaid_content)
print(f" Mermaid guardado en: {mermaid_path}")
```

Puedes usar Mermaid en:
- GitHub Markdown
- GitLab
- Notion
- Mermaid Live Editor

### SVG

```
# Renderizar directamente a SVG
svg_bytes = renderer.render(graph, 'svg', {
    'width': 800,
    'height': 600,
    'background': 'white'
})

svg_path = output_dir / 'graph.svg'
svg_path.write_bytes(svg_bytes)
print(f" SVG guardado en: {svg_path}")
```

### PNG

```
# Renderizar a PNG
png_bytes = renderer.render(graph, 'png', {
    'width': 1200,
    'height': 900,
    'dpi': 150
})

png_path = output_dir / 'graph.png'
png_path.write_bytes(png_bytes)
print(f" PNG guardado en: {png_path}")
```

### JSON

```
# Exportar estructura como JSON
import json

json_data = graph.to_json()
json_str = json.dumps(json_data, indent=2)

print("JSON Output:")
print(json_str[:200] + "...")
print()

json_path = output_dir / 'graph.json'
json_path.write_text(json_str)
print(f" JSON guardado en: {json_path}")
```

## 6. Estilos y Personalización

### Temas Predefinidos

```
themes = ['default', 'dark', 'colorful', 'minimal', 'academic']

for theme in themes:
    themed_graph = graph_generator.generate_graph(
        nodes,
        edges,
        options={'theme': theme}
    )
    
    # Renderizar
    svg = renderer.render(themed_graph, 'svg', {'title': f'Theme: {theme}'})
    path = output_dir / f'graph_theme_{theme}.svg'
    path.write_bytes(svg)
    
    print(f"✓ Theme '{theme}' exportado")
```

### Estilo Personalizado

```
custom_style = {
    'node': {
        'shape': 'rounded_rect',
        'fill': '#E8F4F8',
        'stroke': '#2C3E50',
        'stroke_width': 2,
        'font_family': 'Arial',
        'font_size': 14
    },
    'edge': {
        'stroke': '#7F8C8D',
        'stroke_width': 1.5,
        'arrow_size': 8
    },
    'label': {
        'font_family': 'Courier New',
        'font_size': 12,
        'color': '#34495E'
    }
}

styled_graph = graph_generator.generate_graph(
    nodes,
    edges,
    options={'custom_style': custom_style}
)

print(" Estilo personalizado aplicado")
```

## 7. Exportación Batch

### Exportar en Múltiples Formatos

```
from app.core.visualization.batch_exporter import BatchExporter, ExportTask, ExportFormat, BatchExportConfig, ProcessingMode

# Preparar tareas
formats = [ExportFormat.SVG, ExportFormat.PNG, ExportFormat.DOT, ExportFormat.MERMAID, ExportFormat.JSON]

tasks = [
    ExportTask(
        id=f'social_graph_{fmt.value}',
        visualization_type='graph',
        data={
            'nodes': social_nodes,
            'edges': social_edges
        },
        format=fmt,
        output_path=output_dir / f'social_graph.{fmt.value}'
    )
    for fmt in formats
]

# Configurar batch
config = BatchExportConfig(
    mode=ProcessingMode.ASYNC,
    max_workers=3,
    optimize_output=True
)

exporter = BatchExporter(config)

# Exportar
results = await exporter.export_batch(tasks)

print(f"\n Batch export completado:")
for result in results:
    if result.success:
        size_kb = result.file_size / 1024
        print(f"  ✓ {result.output_path.name} ({size_kb:.1f} KB)")
    else:
        print(f"  ✗ {result.task_id}: {result.error}")
```

## 8. Casos de Uso Avanzados

### Grafo de Dependencias

```
# Sistema de módulos
modules = [
    {'id': 'core', 'label': 'Core', 'level': 0},
    {'id': 'utils', 'label': 'Utils', 'level': 0},
    {'id': 'parser', 'label': 'Parser', 'level': 1},
    {'id': 'analyzer', 'label': 'Analyzer', 'level': 2},
    {'id': 'api', 'label': 'API', 'level': 3},
]

dependencies = [
    ('parser', 'core', {'type': 'depends'}),
    ('parser', 'utils', {'type': 'depends'}),
    ('analyzer', 'parser', {'type': 'depends'}),
    ('analyzer', 'core', {'type': 'depends'}),
    ('api', 'analyzer', {'type': 'depends'}),
    ('api', 'core', {'type': 'depends'}),
]

dep_graph = graph_generator.generate_graph(
    modules,
    dependencies,
    options={
        'directed': True,
        'layout': 'hierarchical',
        'group_by': 'level'
    }
)

print("Grafo de dependencias generado")
```

### Visualización de Estado de Máquina

```
# Máquina de estados finita
states = [
    {'id': 'idle', 'label': 'Idle', 'type': 'initial'},
    {'id': 'processing', 'label': 'Processing', 'type': 'active'},
    {'id': 'success', 'label': 'Success', 'type': 'final'},
    {'id': 'error', 'label': 'Error', 'type': 'final'},
]

transitions = [
    ('idle', 'processing', {'event': 'start', 'label': 'start'}),
    ('processing', 'success', {'event': 'complete', 'label': 'complete'}),
    ('processing', 'error', {'event': 'fail', 'label': 'fail'}),
    ('error', 'idle', {'event': 'retry', 'label': 'retry'}),
    ('success', 'idle', {'event': 'reset', 'label': 'reset'}),
]

fsm_graph = graph_generator.generate_graph(
    states,
    transitions,
    options={
        'directed': True,
        'show_labels': True,
        'style': 'state_machine'
    }
)

print("FSM visualizado")
```

## 9. Optimización para Grafos Grandes

```
from app.core.visualization.optimizer import VisualizationOptimizer, OptimizationConfig, OptimizationLevel

# Crear grafo grande
large_nodes = [{'id': f'n{i}', 'label': f'N{i}'} for i in range(500)]
large_edges = [
    (f'n{i}', f'n{i+1}', {}) for i in range(499)
] + [
    (f'n{i}', f'n{i+10}', {}) for i in range(490)
]

print(f"Grafo grande: {len(large_nodes)} nodos, {len(large_edges)} aristas")

# Optimizar
optimizer = VisualizationOptimizer(
    OptimizationConfig(level=OptimizationLevel.AGGRESSIVE)
)

opt_nodes, opt_edges, metadata = optimizer.optimize_graph(
    large_nodes,
    large_edges
)

print(f"\nOptimizado: {len(opt_nodes)} nodos, {len(opt_edges)} aristas")
print(f"Reducción: {metadata['reduction_ratio']['nodes']*100:.1f}%")
print(f"Optimizaciones: {', '.join(metadata['optimizations_applied'])}")
```

## 10. Comparación de Rendimiento

```
import time
import matplotlib.pyplot as plt

# Medir tiempo de generación vs tamaño
sizes = [10, 50, 100, 200, 500, 1000]
times = []

for size in sizes:
    nodes = [{'id': f'n{i}', 'label': str(i)} for i in range(size)]
    edges = [(f'n{i}', f'n{i+1}', {}) for i in range(size-1)]
    
    start = time.time()
    graph = graph_generator.generate_graph(nodes, edges)
    elapsed = time.time() - start
    
    times.append(elapsed * 1000)  # ms
    print(f"Size {size}: {elapsed*1000:.2f}ms")

# Graficar
plt.figure(figsize=(10, 6))
plt.plot(sizes, times, marker='o', linewidth=2, markersize=8)
plt.xlabel('Número de Nodos')
plt.ylabel('Tiempo de Generación (ms)')
plt.title('Rendimiento de Generación de Grafos')
plt.grid(True, alpha=0.3)
plt.savefig(output_dir / 'performance.png', dpi=150, bbox_inches='tight')
print(f"\n Gráfica de rendimiento guardada")
```

### Mejores Prácticas

1. **Elección de formato**:
   - SVG: Para web y escalabilidad
